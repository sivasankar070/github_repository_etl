```python
from pathlib import Path
import json
import pandas as pd
```

```python
BASE_DIR = Path.cwd().parent

RAW_FILE = BASE_DIR / "data" / "raw" / "repositories.json"

with open(RAW_FILE, 'r', encoding='utf-8') as f:
    data = json.load(f)

df = pd.DataFrame(data)


columns = ["id","name","full_name","description","html_url","private","fork","created_at","updated_at","pushed_at","size","stargazers_count","forks_count","open_issues_count","language","topics","visibility","archived","disabled","default_branch"]

df = df[columns]
```

```python
a = df["topics"].apply(lambda x: not isinstance(x, list))
print(f"Found {a.sum()} non-list values in the topics column out of {len(df)} total rows.")
```

```python
s = 0
for topic in df["topics"]:
    if isinstance(topic, list):
        for item in topic:
            if not isinstance(item, str):
                s += 1
print(f"found {s} non-string value in topics list")
```

```python
s = 0
for idx, topic in df["topics"].items():
    if isinstance(topic, list):
        dup_topic = set()
        for item in topic:
            if item in dup_topic:
                s += 1
                print(f"duplicate topic found in \n id : {df.loc[idx, 'id']} \n name : {df.loc[idx, 'name']} \n topics : {topic}")
            else:
                dup_topic.add(item)
print(f"found {s} duplicate topic values in topics list")
```

```python
unique_topics = set()
for topic in df["topics"]:
    if isinstance(topic, list):
        for item in topic:
            if item.capitalize() == item:
                print(f"Inconsistan capitalization found in topics list: {item}")
            if item.strip() != item:
                print(f"Leading or trailing whitespace found in topics list: '{item}'")
            if item not in unique_topics:
                unique_topics.add(item)
print(f"unique topic in topics lists: {unique_topics}")
```
